# Generación de SBOM

## ¿Qué es un SBOM?

Un **Software Bill of Materials (SBOM)** es un inventario completo de todos los componentes,
bibliotecas y dependencias que conforman una aplicación de software.

### ¿Por qué es importante?
- 📋 **Transparencia**: Conocer exactamente qué contiene tu software
- 🔒 **Seguridad**: Identificar componentes vulnerables rápidamente
- 📜 **Cumplimiento**: Requerido por regulaciones (Orden Ejecutiva 14028 de EE.UU.)
- 🔄 **Gestión de riesgos**: Responder rápido ante nuevas vulnerabilidades

### Herramienta: Syft
Syft (por Anchore) es una herramienta CLI para generar SBOMs en múltiples formatos:
- SPDX
- CycloneDX
- JSON nativo de Syft

```python
import json
import pandas as pd
from pathlib import Path

# Cargar SBOMs generados
results_dir = Path("../data/results")
sbom_files = list(results_dir.glob("*-sbom.json"))

print(f"SBOMs encontrados: {len(sbom_files)}")

for f in sbom_files:
    with open(f) as file:
        data = json.load(file)
        artifacts = data.get("artifacts", [])
        print(f"\n📦 {f.stem.replace('-sbom', '')}")
        print(f"   Componentes detectados: {len(artifacts)}")
        
        if artifacts:
            # Mostrar los primeros 10 componentes
            print(f"   Primeros componentes:")
            for art in artifacts[:10]:
                name = art.get("name", "N/A")
                version = art.get("version", "N/A")
                pkg_type = art.get("type", "N/A")
                print(f"     - {name}@{version} ({pkg_type})")
```
