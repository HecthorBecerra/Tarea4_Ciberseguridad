# Introducción a Vulnerabilidades

## ¿Qué es una vulnerabilidad?

Una vulnerabilidad es una debilidad en el código, configuración o infraestructura
que puede ser explotada por un atacante.

### Tipos de vulnerabilidades:
- **Injection**: Insertar código malicioso
- **Broken Authentication**: Fallos en autenticación
- **Sensitive Data Exposure**: Exposición de datos sensibles
- **XXE**: External XML Entity
- **Broken Access Control**: Control de acceso inadecuado
