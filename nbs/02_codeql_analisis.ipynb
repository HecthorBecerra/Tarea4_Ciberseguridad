# Análisis con CodeQL

## ¿Qué es CodeQL?

CodeQL es una herramienta de análisis semántico de código desarrollada por GitHub.
Permite encontrar vulnerabilidades de seguridad tratando el código como datos consultables.

### Características principales:
- Análisis estático profundo
- Soporte para múltiples lenguajes (Python, JavaScript, Java, C++, C#)
- Consultas personalizables
- Integración con GitHub Actions

```python
import json
from pathlib import Path

# Cargar resultados de CodeQL
results_dir = Path("../data/results")
codeql_files = list(results_dir.glob("*-codeql.json"))

print(f"Archivos de CodeQL encontrados: {len(codeql_files)}")

for f in codeql_files:
    with open(f) as file:
        data = json.load(file)
        print(f"\n📄 {f.name}")
        print(f"   Resultados: {len(data) if isinstance(data, list) else 'N/A'}")
```
